# How to create a Virtual Format file (.vrt)
There are numerous effective formats for storing large remote sensing data as explained in the lecture. Most offer the possibility of storing the data in a chunked or pieced manner, others make reading possible without changing the storage condition. 

One example is the VRT (Virtual Raster) format which does not store data by itself but refers to the collection of image data stored in conventional formats such as tif, gpeg, png, etc. Thus it is said that VRT Raster is a gateway format that uses other GDAL formats. VRT Raster is an XML format that maps its attributes and geometries to that of an underlying data source of any GDAL-supported raster format. 

**Import the necessary packages**

```python
import os, glob, pathlib
import xarray as xr
from osgeo import gdal
```

**Define the path to the folder that contains the images**

```python
# Replace string with your own foloder path
DIR_PATH = pathlib.Path().resolve()
DATA_PATH = DIR_PATH / "data"
os.getcwdb()
```

```python
# Find all tif files in folder
tif_files = glob.glob(os.path.join(DATA_PATH, "*.tif"))
print(f"Found {len(tif_files)} tif files in current folder")
tif_files
```

**Create a composite multiband image in the gdal vrt format using the images in the folder.**

```python
# Define output path and name for the vrt file 
outvrt = os.path.join(DIR_PATH / "data", f'T32VNM_20220517T103631.vrt')
# Build the vrt
gb = gdal.BuildVRT(outvrt, tif_files, separate=True) # Note separate=True places each file in a separate band
gb = None
```

```python
# Open the newly created vrt file with xarray to look at the properties of the image
ds = xr.open_dataset(outvrt)
ds
```
